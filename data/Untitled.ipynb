```python
import requests
import pandas as pd
from time import sleep

# Load existing CSV to get the current count
try:
    existing_df = pd.read_csv("fpl_teams.csv")
    print(f"Loaded existing CSV with {len(existing_df)} records")
except FileNotFoundError:
    existing_df = pd.DataFrame()
    print("No existing CSV found, starting fresh")

team_id = 959412
consecutive_failures = 0
max_consecutive_failures = 2

while True:
    url = f"https://fantasy.premierleague.com/api/entry/{team_id}/"
    response = requests.get(url, headers={"Accept": "application/json"})

    if response.status_code != 200:
        consecutive_failures += 1
        print(f"Failed to fetch team ID {team_id} (attempt {consecutive_failures}/{max_consecutive_failures})")

        if consecutive_failures >= max_consecutive_failures:
            print(f"No more teams found after {max_consecutive_failures} consecutive failures")
            break

        team_id += 1
        continue

    # Reset failure counter on success
    consecutive_failures = 0

    data = response.json()
    team_record = {
        "id": data.get("id"),
        "team_name": data.get("name"),
        "first_name": data.get("player_first_name"),
        "last_name": data.get("player_last_name")
    }

    # Append single record to CSV immediately
    df_new_record = pd.DataFrame([team_record])
    df_new_record.to_csv("fpl_teams.csv", mode='a', header=not pd.io.common.file_exists("fpl_teams.csv") or team_id == 2, index=False)

    if(data.get("id") % 1000 == 0):
        print(f"Processed team ID: {data.get('id')}")

    team_id += 1
    # sleep(0.001)  # small delay to avoid hitting the API too hard

print("Finished updating CSV with new teams!")
```
